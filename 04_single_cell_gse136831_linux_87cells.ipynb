```python
import scanpy as sc
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from statannotations.Annotator import Annotator

# 1. Prepare a DataFrame containing the scores
# Compare the selected groups in 'leiden_res_0.10'
group_col = "leiden_res_0.10"
score_col = "KEGG_PURINE_METABOLISM_score"

# Extract plotting data
df = pd.DataFrame({
    "Score": adata_all.obs[score_col],
    "Group": adata_all.obs[group_col]
})

# Set Group as categorical and retain the first three observed groups
groups = df["Group"].unique()[:3] 
df = df[df["Group"].isin(groups)]

# 2. Set the plotting palette
# Alternative palette:["#E64B35", "#4DBBD5", "#00A087"]
my_pal = sns.color_palette("nipy_spectral", len(groups)) 

# 3. Create plot
plt.figure(figsize=(6, 6))
ax = sns.violinplot(
    data=df, x="Group", y="Score", 
    palette=my_pal, 
    inner="box",      # Show an inner box plot
    linewidth=1.2,    # Line width
    cut=0             # Limit violins to the observed data range
)

# 4. Add significance annotations
# Define pairwise comparisons
pairs = [(groups[0], groups[1]), (groups[1], groups[2]), (groups[0], groups[2])]

annotator = Annotator(ax, pairs, data=df, x="Group", y="Score")
annotator.configure(
    test='Mann-Whitney',     # Nonparametric comparison of the selected cell-level scores
    text_format='star',      # Display significance stars
    loc='inside', 
    line_width=1,
    color='black'
)
annotator.apply_test().annotate()
# 5. Adjust plot appearance
plt.title("Purine Metabolism Scoring", fontsize=15, pad=20)
plt.xlabel("Cell Groups", fontsize=12, fontweight='bold')
plt.ylabel("Pathway Score", fontsize=12, fontweight='bold')

# Remove the top and right spines
sns.despine()

# Save
plt.tight_layout()
plt.savefig(NOTEBOOK_RESULTS_DIR / "Pathway_Score_Significance_Violin.pdf", dpi=300)
plt.show()
```
