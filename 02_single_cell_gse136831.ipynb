```python
import scanpy as sc
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import sparse

# =========================
# 1. Parameters
# =========================
gene = "SRSF1"
celltype_key = "celltype3"
disease_key = "diseaseType"
disease_order = ["IPF", "CON"]
missing_label = "(Empty/NA)"

# =========================
# 2. Check data
# =========================
if gene not in adata.var_names and (
    adata.raw is None or gene not in adata.raw.var_names
):
    raise ValueError(
        f"{gene} not found in adata.var_names or adata.raw.var_names"
    )

if celltype_key not in adata.obs.columns:
    raise ValueError(f"{celltype_key} not found in adata.obs")

if disease_key not in adata.obs.columns:
    if "diseasetype" in adata.obs.columns:
        disease_key = "diseasetype"
    else:
        raise ValueError(
            "Neither 'diseaseType' nor 'diseasetype' exists in adata.obs"
        )

# =========================
# 3. Extract expression
# =========================
if adata.raw is not None and gene in adata.raw.var_names:
    expr = adata.raw[:, gene].X
else:
    expr = adata[:, gene].X

if sparse.issparse(expr):
    expr = expr.toarray().ravel()
else:
    expr = np.asarray(expr).ravel()

# =========================
# 4. Preserve all cell type labels
# =========================
celltype_series = adata.obs[celltype_key]

# Retain all predefined categorical levels, including unused categories
if isinstance(celltype_series.dtype, pd.CategoricalDtype):
    original_categories = [
        str(x) for x in celltype_series.cat.categories
    ]
else:
    original_categories = [
        str(x) for x in celltype_series.dropna().unique()
    ]

# Replace missing and empty labels with a visible label
def clean_celltype(x):
    if pd.isna(x) or str(x).strip() == "":
        return missing_label
    return str(x)

celltype_values = celltype_series.map(clean_celltype).to_numpy()

# Preserve the original category order
celltype_order = []
for ct in original_categories:
    clean_ct = clean_celltype(ct)
    if clean_ct not in celltype_order:
        celltype_order.append(clean_ct)

# Add categories represented by cells but absent from the original category list
for ct in celltype_values:
    if ct not in celltype_order:
        celltype_order.append(ct)

# Make sure the missing/empty label is retained when present
has_missing_or_empty = (
    celltype_series.isna().any()
    or any(str(x).strip() == "" for x in celltype_series.dropna())
    or any(str(x).strip() == "" for x in original_categories)
)

if has_missing_or_empty and missing_label not in celltype_order:
    celltype_order.append(missing_label)

# =========================
# 5. Build plotting dataframe
# =========================
disease_values = (
    adata.obs[disease_key]
    .astype("string")
    .str.strip()
    .to_numpy()
)

df = pd.DataFrame({
    "Cell type": celltype_values,
    "Disease": disease_values,
    "Expression": expr
})

# Keep IPF and CON only; do not remove zero-expression cells
df = df[df["Disease"].isin(disease_order)].copy()

df["Cell type"] = pd.Categorical(
    df["Cell type"],
    categories=celltype_order,
    ordered=True
)

df["Disease"] = pd.Categorical(
    df["Disease"],
    categories=disease_order,
    ordered=True
)

# =========================
# 6. Print cell counts
# =========================
print("Cell counts by cell type and disease:")
print(
    pd.crosstab(
        df["Cell type"],
        df["Disease"],
        dropna=False
    ).reindex(
        index=celltype_order,
        columns=disease_order,
        fill_value=0
    )
)

print(f"\nTotal cells plotted: {len(df):,}")
print(f"Total celltype3 labels retained: {len(celltype_order)}")
print(f"Zero-expression cells retained: {(df['Expression'] == 0).sum():,}")

# =========================
# 7. Plot violin plots
# =========================
sns.set_style("white")

fig_width = max(8, len(celltype_order) * 0.55)
fig, ax = plt.subplots(figsize=(fig_width, 4.8))

palette = {
    "IPF": "#FA7F6F",
    "CON": "#82B0D2"
}

sns.violinplot(
    data=df,
    x="Cell type",
    y="Expression",
    hue="Disease",
    order=celltype_order,
    hue_order=disease_order,
    palette=palette,
    dodge=True,
    inner=None,
    cut=0,
    linewidth=0.6,
    width=0.85,
    ax=ax
)

# =========================
# 8. Formatting
# =========================
ax.set_title(gene, fontsize=13, pad=8)
ax.set_xlabel("")
ax.set_ylabel("Expression Level", fontsize=12)

ax.tick_params(
    axis="x",
    labelrotation=55,
    labelsize=9,
    length=0
)
ax.tick_params(
    axis="y",
    labelsize=10,
    direction="out",
    length=3
)

for label in ax.get_xticklabels():
    label.set_horizontalalignment("right")

ax.legend(
    title="Disease",
    frameon=False,
    loc="upper right"
)

sns.despine(ax=ax, top=True, right=True)
ax.grid(False)

plt.tight_layout()

# =========================
# 9. Save figures
# =========================
fig.savefig(
    NOTEBOOK_RESULTS_DIR / f"{gene}_violin_celltype3_IPF_vs_CON.pdf",
    bbox_inches="tight",
    dpi=300
)

fig.savefig(
    NOTEBOOK_RESULTS_DIR / f"{gene}_violin_celltype3_IPF_vs_CON.png",
    bbox_inches="tight",
    dpi=300
)

plt.show()
```

```python

import scanpy as sc
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from scipy import sparse
from scipy.stats import mannwhitneyu
from statsmodels.stats.multitest import multipletests

# =========================
# 1. Parameters
# =========================
gene = "SRSF1"
celltype_key = "celltype3"
disease_key = "diseaseType"
disease_order = ["IPF", "CON"]

# Statistical parameters
min_cells = 3
p_adjust_method = "fdr_bh"

# =========================
# 2. Check data
# =========================
if gene not in adata.var_names and (
    adata.raw is None or gene not in adata.raw.var_names
):
    raise ValueError(f"{gene} not found in adata.var_names or adata.raw.var_names")

if celltype_key not in adata.obs.columns:
    raise ValueError(f"{celltype_key} not found in adata.obs")

if disease_key not in adata.obs.columns:
    if "diseasetype" in adata.obs.columns:
        disease_key = "diseasetype"
    else:
        raise ValueError("Neither diseaseType nor diseasetype exists")

# =========================
# 3. Extract expression
# =========================
if adata.raw is not None and gene in adata.raw.var_names:
    expr = adata.raw[:, gene].X
else:
    expr = adata[:, gene].X

if sparse.issparse(expr):
    expr = expr.toarray().ravel()
else:
    expr = np.asarray(expr).ravel()

df = pd.DataFrame({
    "Cell type": adata.obs[celltype_key].astype(str).to_numpy(),
    "Disease": adata.obs[disease_key].astype(str).to_numpy(),
    "Expression": expr
})

# Keep IPF and CON
df = df[df["Disease"].isin(disease_order)].copy()

# Remove zero-expression cells
df = df[df["Expression"] > 0].copy()

# =========================
# 4. Preserve cell type order
# =========================
if isinstance(adata.obs[celltype_key].dtype, pd.CategoricalDtype):
    celltype_order = [
        str(ct) for ct in adata.obs[celltype_key].cat.categories
        if str(ct) in df["Cell type"].values
    ]
else:
    celltype_order = [
        ct for ct in adata.obs[celltype_key].astype(str).unique()
        if ct in df["Cell type"].values
    ]

df["Cell type"] = pd.Categorical(
    df["Cell type"],
    categories=celltype_order,
    ordered=True
)

df["Disease"] = pd.Categorical(
    df["Disease"],
    categories=disease_order,
    ordered=True
)

print("Positive cells by disease and cell type:")
print(pd.crosstab(df["Cell type"], df["Disease"]))
print(f"\nTotal positive cells: {len(df):,}")

# =========================
# 5. Statistical analysis
# =========================
results = []

for ct in celltype_order:

    ipf = df.loc[
        (df["Cell type"] == ct) &
        (df["Disease"] == "IPF"),
        "Expression"
    ].values

    con = df.loc[
        (df["Cell type"] == ct) &
        (df["Disease"] == "CON"),
        "Expression"
    ].values

    if len(ipf) >= min_cells and len(con) >= min_cells:

        stat, pvalue = mannwhitneyu(
            ipf,
            con,
            alternative="two-sided"
        )

        results.append({
            "Cell type": ct,
            "N_IPF": len(ipf),
            "N_CON": len(con),
            "Mean_IPF": np.mean(ipf),
            "Mean_CON": np.mean(con),
            "Median_IPF": np.median(ipf),
            "Median_CON": np.median(con),
            "P_value": pvalue
        })

    else:
        results.append({
            "Cell type": ct,
            "N_IPF": len(ipf),
            "N_CON": len(con),
            "Mean_IPF": np.mean(ipf) if len(ipf) else np.nan,
            "Mean_CON": np.mean(con) if len(con) else np.nan,
            "Median_IPF": np.median(ipf) if len(ipf) else np.nan,
            "Median_CON": np.median(con) if len(con) else np.nan,
            "P_value": np.nan
        })

stats_df = pd.DataFrame(results)

# Multiple testing correction
valid = stats_df["P_value"].notna()

stats_df["P_adj"] = np.nan

if valid.sum() > 0:
    stats_df.loc[valid, "P_adj"] = multipletests(
        stats_df.loc[valid, "P_value"],
        method=p_adjust_method
    )[1]

# Convert P values to significance stars
def get_significance(p):
    if pd.isna(p):
        return ""
    elif p < 0.001:
        return "***"
    elif p < 0.01:
        return "**"
    elif p < 0.05:
        return "*"
    else:
        return ""

stats_df["Significance"] = stats_df["P_adj"].apply(get_significance)

print("\nStatistical results:")
print(stats_df.to_string(index=False))

# Save statistical results
stats_df.to_csv(
    NOTEBOOK_RESULTS_DIR / f"{gene}_IPF_vs_CON_statistics.csv",
    index=False
)

# =========================
# 6. Plot violin plots
# =========================
sns.set_style("white")

fig_width = max(6, len(celltype_order) * 0.45)

fig, ax = plt.subplots(figsize=(fig_width, 4.8))

palette = {
    "IPF": "#FA7F6F",
    "CON": "#82B0D2"
}

sns.violinplot(
    data=df,
    x="Cell type",
    y="Expression",
    hue="Disease",
    order=celltype_order,
    hue_order=disease_order,
    palette=palette,
    dodge=True,
    inner=None,
    cut=0,
    linewidth=0.6,
    width=0.85,
    ax=ax
)

# =========================
# 7. Add significance brackets
# =========================
y_min, y_max = ax.get_ylim()
y_range = y_max - y_min

for i, ct in enumerate(celltype_order):

    row = stats_df.loc[stats_df["Cell type"] == ct].iloc[0]

    stars = row["Significance"]

    # Only display significant results
    if stars == "":
        continue

    ipf = df.loc[
        (df["Cell type"] == ct) &
        (df["Disease"] == "IPF"),
        "Expression"
    ]

    con = df.loc[
        (df["Cell type"] == ct) &
        (df["Disease"] == "CON"),
        "Expression"
    ]

    if len(ipf) == 0 or len(con) == 0:
        continue

    # Positions of the two violins
    x1 = i - 0.85 / 4
    x2 = i + 0.85 / 4

    # Bracket height
    local_max = max(ipf.max(), con.max())

    y = local_max + y_range * 0.035
    h = y_range * 0.012

    # Draw bracket
    ax.plot(
        [x1, x1, x2, x2],
        [y, y + h, y + h, y],
        lw=0.9,
        c="black"
    )

    # Add significance stars
    ax.text(
        (x1 + x2) / 2,
        y + h + y_range * 0.005,
        stars,
        ha="center",
        va="bottom",
        fontsize=10,
        fontweight="bold"
    )

# Expand y-axis to prevent clipping
sig_rows = stats_df[stats_df["Significance"] != ""]

if len(sig_rows) > 0:
    sig_max = max(
        max(
            df.loc[
                (df["Cell type"] == ct) &
                (df["Disease"].isin(disease_order)),
                "Expression"
            ].max()
            for ct in sig_rows["Cell type"]
        ),
        0
    )

    ax.set_ylim(
        y_min,
        max(y_max, sig_max + y_range * 0.12)
    )

# =========================
# 8. Formatting
# =========================
ax.set_title(gene, fontsize=13, pad=8)
ax.set_xlabel("")
ax.set_ylabel("Expression Level", fontsize=12)

ax.tick_params(
    axis="x",
    labelrotation=55,
    labelsize=9,
    length=0
)

ax.tick_params(
    axis="y",
    labelsize=10,
    direction="out",
    length=3
)

for label in ax.get_xticklabels():
    label.set_horizontalalignment("right")

ax.legend(
    title="Disease",
    frameon=False,
    loc="upper right"
)

sns.despine(ax=ax, top=True, right=True)
ax.grid(False)

plt.tight_layout()

# =========================
# 9. Save figures
# =========================
fig.savefig(
    NOTEBOOK_RESULTS_DIR / f"{gene}_violin_celltype3_IPF_vs_CON.pdf",
    bbox_inches="tight",
    dpi=300
)

fig.savefig(
    NOTEBOOK_RESULTS_DIR / f"{gene}_violin_celltype3_IPF_vs_CON.png",
    bbox_inches="tight",
    dpi=300
)

plt.show()
```

```python
import scanpy as sc
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from statannotations.Annotator import Annotator

# 1. Prepare a DataFrame containing the scores
# Compare the selected groups in 'leiden_res_0.10'
group_col = "leiden_res_0.10"
score_col = "KEGG_PURINE_METABOLISM_score"

# Extract plotting data
df = pd.DataFrame({
    "Score": adata_all.obs[score_col],
    "Group": adata_all.obs[group_col]
})

# Set Group as categorical and retain the first three observed groups
groups = df["Group"].unique()[:3] 
df = df[df["Group"].isin(groups)]

# 2. Set the plotting palette
# Alternative palette:["#E64B35", "#4DBBD5", "#00A087"]
my_pal = sns.color_palette("nipy_spectral", len(groups)) 

# 3. Create plot
plt.figure(figsize=(6, 6))
ax = sns.violinplot(
    data=df, x="Group", y="Score", 
    palette=my_pal, 
    inner="box",      # Show an inner box plot
    linewidth=1.2,    # Line width
    cut=0             # Limit violins to the observed data range
)

# 4. Add significance annotations
# Define pairwise comparisons
pairs = [(groups[0], groups[1]), (groups[1], groups[2]), (groups[0], groups[2])]

annotator = Annotator(ax, pairs, data=df, x="Group", y="Score")
annotator.configure(
    test='Mann-Whitney',     # Nonparametric comparison of the selected cell-level scores
    text_format='star',      # Display significance stars
    loc='inside', 
    line_width=1,
    color='black'
)
annotator.apply_test().annotate()
# 5. Adjust plot appearance
plt.title("Purine Metabolism Scoring", fontsize=15, pad=20)
plt.xlabel("Cell Groups", fontsize=12, fontweight='bold')
plt.ylabel("Pathway Score", fontsize=12, fontweight='bold')

# Remove the top and right spines
sns.despine()

# Save
plt.tight_layout()
plt.savefig(NOTEBOOK_RESULTS_DIR / "Pathway_Score_Significance_Violin.pdf", dpi=300)
plt.show()
```
